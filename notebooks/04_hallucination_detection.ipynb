# Hallucination Detection Analysis

Analyze engineering hallucinations generated by AI models.

```python
import pandas as pd

results = pd.read_csv(
    "../data/processed/evaluation_results.csv"
)

results.head()
```

```python
hallucinations = (
    results["hallucination_level"]
    .value_counts()
)

hallucinations
```

```python
import matplotlib.pyplot as plt

hallucinations.plot(
    kind="bar"
)

plt.title(
    "Hallucination Frequency"
)

plt.ylabel("Count")
plt.show()
```

```python
critical_cases = results[
    results["hallucination_level"]
    == "Critical"
]

critical_cases
```

```python
critical_cases.groupby(
    "category"
).size()
```
