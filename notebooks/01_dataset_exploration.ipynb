# Dataset Exploration

This notebook explores the Electrical Engineering AI Evaluation Benchmark dataset.

Objectives:

- Inspect dataset structure
- Analyze category distribution
- Analyze difficulty levels
- Verify dataset quality

```python
import pandas as pd

questions = pd.read_csv("../data/raw/electrical_qa_dataset.csv")

questions.head()
```

```python
questions.info()
```

```python
questions["category"].value_counts()
```

```python
import matplotlib.pyplot as plt

questions["category"].value_counts().plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Question Distribution by Category")
plt.ylabel("Count")
plt.show()
```

```python
questions["difficulty"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Difficulty Distribution")
plt.show()
```
