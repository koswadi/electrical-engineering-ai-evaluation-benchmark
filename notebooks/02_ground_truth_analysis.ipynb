# Ground Truth Analysis

This notebook analyzes the reference answers used for AI evaluation.

```python
import pandas as pd

ground_truth = pd.read_csv(
    "../data/raw/ground_truth_answers.csv"
)

ground_truth.head()
```

```python
ground_truth["answer_length"] = (
    ground_truth["ground_truth"]
    .str.split()
    .apply(len)
)

ground_truth["answer_length"].describe()
```

```python
import matplotlib.pyplot as plt

ground_truth["answer_length"].hist(
    bins=10
)

plt.title("Ground Truth Answer Length")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.show()
```

```python
ground_truth.sort_values(
    "answer_length",
    ascending=False
).head()
```
