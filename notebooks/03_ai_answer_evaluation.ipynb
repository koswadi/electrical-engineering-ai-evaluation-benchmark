# AI Answer Evaluation

Compare AI-generated answers against engineering ground truth answers.

```python
import pandas as pd

benchmark = pd.read_csv(
    "../data/processed/benchmark_dataset.csv"
)

benchmark.head()
```

```python
evaluation = pd.read_csv(
    "../data/processed/evaluation_results.csv"
)

evaluation.head()
```

```python
evaluation.groupby(
    "model_name"
)["technical_accuracy_score"].mean()
```

```python
evaluation.groupby(
    "model_name"
)["similarity_score"].mean()
```

```python
import matplotlib.pyplot as plt

evaluation.groupby(
    "model_name"
)["technical_accuracy_score"].mean().plot(
    kind="bar"
)

plt.title("Average Technical Accuracy")
plt.ylabel("Score")
plt.show()
```
