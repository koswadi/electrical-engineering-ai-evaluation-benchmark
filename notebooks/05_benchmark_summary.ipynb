# Benchmark Summary Report

Generate benchmark performance statistics and summary visualizations.

```python
import pandas as pd

results = pd.read_csv(
    "../data/processed/evaluation_results.csv"
)

results.head()
```

```python
summary = results.groupby(
    "model_name"
).agg({
    "technical_accuracy_score":"mean",
    "similarity_score":"mean"
})

summary
```

```python
summary.plot(
    kind="bar",
    figsize=(8,5)
)
```

```python
hallucination_rate = (
    results[
        results["hallucination_level"]
        != "None"
    ]
    .groupby("model_name")
    .size()
)

hallucination_rate
```

```python
import matplotlib.pyplot as plt

hallucination_rate.plot(
    kind="bar"
)

plt.title(
    "Hallucination Rate by Model"
)

plt.ylabel("Cases")
plt.show()
```

```python
results.groupby(
    "category"
)["technical_accuracy_score"].mean().sort_values(
    ascending=False
)
```
